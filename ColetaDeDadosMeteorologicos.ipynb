```python
# https://docs.google.com/spreadsheets/d/14DGPDTaqlNiwq5WcXFmaialxj9flEbz2ijo-hbHNomo/edit?usp=sharing

import pandas as pd
import requests
import time

#SEMPRE QUE FOR FAZER PLANILHAS UTILIZAR O COMANDO - export?format=csv

endpoint_planilha = "https://docs.google.com/spreadsheets/d/14DGPDTaqlNiwq5WcXFmaialxj9flEbz2ijo-hbHNomo/export?format=csv"
endpoint_api = "https://api.open-meteo.com/v1/forecast"
```

```python
# LER OS DADOS COLETADOS E EXIBIR NO PYTHON

print(" === Dados Coletados === ")
dados_planilha = pd.read_csv(endpoint_planilha)

display(dados_planilha)
```

```python
# Iterar cada resposta do formulário

# O zero na tabela indica posição - no caso o correto é ÍNDICE/INDEX
# iterrows() faz com que leia uma linha por vez da base de dados

for indice, linha in dados_planilha.iterrows():
  nome_local = linha["Nome do local"]
  lat = linha["Latitude"]
  lon = linha ["Longitude"]

  parametros = {
      'latitude' : lat,
      'longitude': lon,
      'current' : 'temperature_2m'
  }

  # O python por usar o FOR, lê linha por linha do excel e traz parametro da api

  resposta = requests.get(endpoint_api, params=parametros).json()

  display(resposta)
  print('\n\n')
```
